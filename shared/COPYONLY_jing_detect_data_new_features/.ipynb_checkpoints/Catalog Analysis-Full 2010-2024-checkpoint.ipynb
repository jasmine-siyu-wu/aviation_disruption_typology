```python
catalog['StartMM'] = catalog['StartYYYYMM']%100
catalog.groupby('StartMM').sum()['Duration']
```

```python
# Improve label clarity by using larger font sizes and adding a legend
labels = [
    "Spring (Month 3-5)",
    "Summer (Month 6-8)",
    "Fall (Month 9-11)",
    "Winter (Month 12-2)",
]
sizes = [102,179,33,141]

fig, ax = plt.subplots(figsize=(7, 7))
wedges, texts, autotexts = ax.pie(
    sizes, 
    autopct='%1.0f%%', 
    startangle=90,
    wedgeprops=dict(width=0.5, edgecolor='w')
)

# Add legend for clarity
ax.legend(
    wedges, labels,
    title="Season",
    loc="upper center",
    bbox_to_anchor=(0.8,0.1),  # move below the pie
    fontsize=12,
    ncol=4                       # number of columns → all labels in one line
)


#ax.set_title("Reasons for Disruptive Events", fontsize=14)
plt.show()
```

```python
catalog['StartYY'] = catalog['StartYYYYMM']//100
catalog.groupby('StartYY').sum()['Duration']
```

```python
catalog['StartYY'] = catalog['StartYYYYMM']//100
catalog.groupby('StartYY').count()['Duration']
```

```python
catalog = catalog[catalog['StartYY']!=2020]
catalog = catalog[catalog['StartYY']!=2021]
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10,3))

catalog_by_year = catalog.groupby('StartYY').agg(
    Duration_sum=('Duration', 'sum'),
    Duration_count=('Duration', 'count')
)
catalog_by_year.iloc[-1] = [48,29]
ax.plot(catalog_by_year.index, catalog_by_year['Duration_sum'], label='Disrupted Days')
ax.plot(catalog_by_year.index, catalog_by_year['Duration_count'], label='Disrupted Events')

ax.legend()
ax.grid(axis="y")

# Force x-axis to show every year
ax.set_xticks(catalog_by_year.index,[2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2022,2023,2024])

plt.show()
```
