```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
combine = pd.read_csv('combinedResults.csv')
```

```python
combine
```

```python
final = combine.copy()
final['DEP_YYYYMMM'] = combine['YYYYMM']
final['month']=combine['YYYYMM']%100
final['year']=combine['YYYYMM']//100
```

```python
final.groupby('year').mean('anomaly_scores_scaled')['anomaly_scores_scaled']
```

```python
year = pd.DataFrame([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],columns = ['year'])
year.set_index(['year'],inplace=True)
year['value'] = final[(final['labels']==5)].groupby('year').count()['DEP_YYYYMMM']
year = year.fillna(0)
```

```python
year1 = pd.DataFrame([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],columns = ['year'])
year1.set_index(['year'],inplace=True)
year1['value'] = final[(final['labels']==3)].groupby('year').count()['DEP_YYYYMMM']
year1 = year1.fillna(0)
year2 = pd.DataFrame([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],columns = ['year'])
year2.set_index(['year'],inplace=True)
year2['value'] = final[(final['labels']==6)].groupby('year').count()['DEP_YYYYMMM']
year2 = year2.fillna(0)
year3 = pd.DataFrame([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],columns = ['year'])
year3.set_index(['year'],inplace=True)
year3['value'] = final[(final['labels']==10)].groupby('year').count()['DEP_YYYYMMM']
year3 = year2.fillna(0)
```

```python
fig, ax = plt.subplots(figsize = (6.9,2.3),dpi = 300,layout="constrained")

# Plot the first set of bars
ax.bar([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024], year['value']/final.groupby('year').count()['DEP_YYYYMMM'], color = '#c82423',label = 'NAS Disruption')
ax.bar([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024], year1['value']/final.groupby('year').count()['DEP_YYYYMMM'],color = '#f8ac8c',  bottom=year['value']/final.groupby('year').count()['DEP_YYYYMMM'],label = 'East Super Disruption')
ax.bar([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024], year2['value']/final.groupby('year').count()['DEP_YYYYMMM'],color = '#2878b5',  bottom=year['value']/final.groupby('year').count()['DEP_YYYYMMM']+year1['value']/final.groupby('year').count()['DEP_YYYYMMM'],label = 'Northeast Disruption')
ax.bar([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024], year3['value']/final.groupby('year').count()['DEP_YYYYMMM'],color = '#9ac9db',  bottom=year['value']/final.groupby('year').count()['DEP_YYYYMMM']+year1['value']/final.groupby('year').count()['DEP_YYYYMMM']+year2['value']/final.groupby('year').count()['DEP_YYYYMMM'],label = 'Southeast Disruption')

plt.xticks([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],
           [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,'2020*','2021*',2022,2023,'2024*'])
plt.legend(loc=2)
plt.ylabel('Ratio of Days')
plt.xlabel('Year')
#plt.savefig("../Figures for ATM 25/Figure11.png", dpi=300, bbox_inches='tight');
```

```python
plt.figure(figsize=(7.5,2.5))
plt.plot([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],(year['value']+year1['value']+year2['value']+year3['value'])/final.groupby('year').count()['DEP_YYYYMMM'],marker = 'o',label = '2 Disrupted Clusters')

#plt.scatter([2020,2021,2024],[3/60*365,14/214*365,24/213*365],marker = 'x')
plt.xticks([2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,2020,2021,2022,2023,2024],
           [2010,2011,2012,2013,2014,2015,2016,2017,2018,2019,'2020*','2021*',2022,2023,'2024*'])
plt.legend()
plt.ylabel('Ratio of Days');
```

```python
year = pd.DataFrame(range(13),columns = ['year'])
year.set_index(['year'],inplace=True)
year['value'] = final[(final['labels']==5)].groupby('month').count()['DEP_YYYYMMM']
year = year.fillna(0)
year1 = pd.DataFrame(range(13),columns = ['year'])
year1.set_index(['year'],inplace=True)
year1['value'] = final[(final['labels']==3)].groupby('month').count()['DEP_YYYYMMM']
year1 = year1.fillna(0)
year2 = pd.DataFrame(range(13),columns = ['year'])
year2.set_index(['year'],inplace=True)
year2['value'] = final[(final['labels']==6)].groupby('month').count()['DEP_YYYYMMM']
year2 = year2.fillna(0)
year3 = pd.DataFrame(range(13),columns = ['year'])
year3.set_index(['year'],inplace=True)
year3['value'] = final[(final['labels']==10)].groupby('month').count()['DEP_YYYYMMM']
year3 = year3.fillna(0)
```

```python
fig, ax = plt.subplots(figsize = (6.9,2.3),dpi = 300,layout="constrained")

# Plot the first set of bars
ax.bar(range(13), year['value']/final.groupby('month').count()['DEP_YYYYMMM'], color = '#c82423',label = 'NAS Disruption')
ax.bar(range(13), year1['value']/final.groupby('month').count()['DEP_YYYYMMM'],color = '#f8ac8c',  bottom=year['value']/final.groupby('month').count()['DEP_YYYYMMM'],label = 'East Super Disruption')
ax.bar(range(13), year2['value']/final.groupby('month').count()['DEP_YYYYMMM'],color = '#2878b5',  bottom=year['value']/final.groupby('month').count()['DEP_YYYYMMM']+year1['value']/final.groupby('month').count()['DEP_YYYYMMM'],label = 'Northeast Disruption')
ax.bar(range(13), year3['value']/final.groupby('month').count()['DEP_YYYYMMM'],color = '#9ac9db',  bottom=year['value']/final.groupby('month').count()['DEP_YYYYMMM']+year1['value']/final.groupby('month').count()['DEP_YYYYMMM']+year2['value']/final.groupby('month').count()['DEP_YYYYMMM'],label = 'Southeast Disruption')

plt.xticks(range(1,13),
           range(1,13))
plt.xlabel('Month')
plt.legend()
plt.ylabel('Ratio of Days')
plt.savefig("../Figures for ATM 25/Figure12.png", dpi=300, bbox_inches='tight');
```

```python
plt.figure(figsize=(7.5,2.5))
plt.plot(range(13),(year['value']+year1['value']+year2['value']+year3['value'])/final.groupby('month').count()['DEP_YYYYMMM'],marker = 'o',label = '3 Disrupted Clusters')

plt.yticks([0,0.03,0.06,0.09,0.12],
           ['0%','3%','6%','9%','12%'])
plt.ylabel('Ratio of Days')
plt.legend()
plt.xticks(range(1,13),range(1,13))
plt.xlabel('Month');
```

```python
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
name = []
for i in huborzone_temp:
    name.append(i+'count')
```

```python
final['count'] = final[name].sum(axis=1)
```

```python
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
name = []
for i in huborzone_temp:
    name.append(i+'cx')
```

```python
final
```

```python
final['cx'] = final[name].sum(axis=1)
```

```python
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
name = []
for i in huborzone_temp:
    name.append(i+'total')
```

```python
for i in huborzone_temp:  
    final[i+'total'] = final[i+'arr_delay']/60*final[i+'count']
```

```python
final['total_arr_delay'] = final[name].sum(axis=1)
```

```python
final['avg'] = final['total_arr_delay']/final['count']
```

```python
results = final.groupby('labels').mean()[['cx','count','avg']]
results['total'] = (results['cx']+results['count'])/2
results['rate'] = results['cx']/(results['total']*2)
```
