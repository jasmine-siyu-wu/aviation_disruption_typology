```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv('final_full_24_25.csv')
```

```python
data2 = pd.read_csv('final_12.csv')
```

```python
name = ['YYYYMM','DAY','Weekday']
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
for i in huborzone_temp:
    name.append(i+'CX_rate')
    name.append(i+'arr_delay')
    name.append(i+'dep_delay')
    name.append(i+'air_delay')
name.append('labels')
```

```python
data = data[name]
data = data[data['YYYYMM']>=202400]
data2 = data2[name]
data2 = data2[data2['YYYYMM']<202408]
```

```python
data24 = pd.concat([data2,data])
```

```python
data24 = data24[(data24['YYYYMM']>=202400)&(data24['YYYYMM']<=202500)]
```

```python
name = []
for i in huborzone_temp:
    name.append(i+'CX_rate')
    name.append(i+'arr_delay')
    name.append(i+'dep_delay')
    name.append(i+'air_delay')
    
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Standardize the data
scaler = StandardScaler()
x_scaled = scaler.fit_transform(data24[name])

# Apply PCA
pca = PCA(n_components=30, random_state=42)
principal_components = pca.fit_transform(x_scaled)

# Create DataFrame for the principal components
pca_df = pd.DataFrame(data=principal_components)


# Print the explained variance
explained_variance = pca.explained_variance_ratio_
explained_variance_sum = sum(explained_variance)
print(explained_variance_sum,f'Explained Variance: {explained_variance}')

loadings = pca.components_

# Create a DataFrame for better readability
loadings_df = pd.DataFrame(loadings, )
```

```python
eigenvalues = pca.explained_variance_
```

```python
sum_explained_variance = []
for i in range(len(explained_variance)):
    sum_explained_variance.append(sum(explained_variance[:i+1])*100)
plt.plot(sum_explained_variance,label = '% Cumulative Variance')
plt.plot(explained_variance*100,label = '% Explained Variance')
plt.plot(eigenvalues,label = 'Eigenvalues')
plt.axhline(y=80,linestyle='dashed',c = 'red',label='y=80')
plt.axhline(y=1,linestyle=':',c = 'red',label='y=1')
plt.xlabel('PC Number')
plt.legend()
#plt.savefig("../Figures for ATM 25/Figure4.png", dpi=300, bbox_inches='tight');
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# Generate synthetic data
#X, _ = make_blobs(n_samples=300, centers=5, cluster_std=0.60, random_state=0)

silhouette_scores = []
ws_scores = []

# Test KMeans with different numbers of clusters
for k in range(2, 21):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(pca_df[:23])
    labels = kmeans.labels_
    score = silhouette_score(pca_df[:23], labels)
    silhouette_scores.append(score)
    ws_scores.append(kmeans.inertia_)

fig, ax1 = plt.subplots(figsize=(8, 6))

# Plot the silhouette scores for different numbers of clusters

line1, = ax1.plot(range(2, 21), silhouette_scores, marker='o',color = 'green',label='Silhouette Score')
plt.legend()
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
ax1.scatter(11,silhouette_scores[9],facecolors='none', edgecolors='red', s=300)
ax2 = ax1.twinx()
line2, = ax2.plot(range(2, 21), ws_scores, alpha=0.5,label = 'WS score')
plt.ylabel('WS Score')

plt.xticks(range(2,21))
#plt.title('Silhouette Score for Optimal Number of Clusters')

lines = [line1, line2]
labels = [line.get_label() for line in lines]
ax1.legend(lines, labels, loc="upper right", frameon=True)  

ax1.grid(True)
#plt.savefig("../Figures for ATM 25/Figure6.png", dpi=300, bbox_inches='tight');
```

```python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data
np.random.seed(42)
X = np.array(pca_df.iloc[:,:23])
#X = np.random.rand(100, 2)

# Perform KMeans clustering
kmeans = KMeans(n_clusters=11, random_state=42)
kmeans.fit(X)

# Get labels and cluster centers
labels = kmeans.labels_
centers = kmeans.cluster_centers_
```

```python
from sklearn.metrics import pairwise_distances_argmin_min
closest, _ = pairwise_distances_argmin_min(centers, pca_df.iloc[:,:23])
```

```python
airportLocation = pd.read_csv('airportLocation.csv').drop('Unnamed: 0',axis=1)
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
L = [' PHX',' LAX',' SAN',' SFO',' DEN',' FLL',' MIA',' MCO',' TPA',' ATL',
     ' MDW',' ORD',' BWI',' BOS',' DTW',' MSP',' LAS',' EWR',' JFK',' LGA',
     ' CLT',' PHL',' BNA',' AUS',' DFW',' IAH',' SLC',' DCA',' IAD',' SEA',' MCI',' IND',' HNL',' ANC']
hubAirportLocation = airportLocation[airportLocation['LOCID'].isin(L)]
order = list(huborzone_temp)
hubAirportLocation['huborzone'] = pd.Categorical(hubAirportLocation['huborzone'], categories=order, ordered=True)
hubAirportLocation = hubAirportLocation.sort_values(by='huborzone')
```

```python
from matplotlib.colors import Normalize
from matplotlib import cm
from matplotlib.colorbar import ColorbarBase
for j in range(11):
    labeledFinal = data24[data24['label_new']==j][name]
    cmap = cm.coolwarm
    fig, ax = plt.subplots(figsize=(3, 3),dpi=300)
    for i in range(len(hubAirportLocation)):
        ax.text(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'], hubAirportLocation.iloc[i]['LOCID'],
            fontsize=6, ha='center', color='black', weight='bold')
    for i in range(34):
        plt.scatter(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'],
                   s = labeledFinal.mean()[i*4]*300,#cx_rate
                   c = cmap((labeledFinal.mean()[i*4+1]+15*60)/(3600)))#arrival delay
    ax.tick_params(axis='x', labelsize=9)
    ax.tick_params(axis='y', labelsize=9)
    plt.xlabel('Lon',fontsize = 9)
    plt.ylabel('Lat',fontsize = 9)
    plt.title('Cluster{}'.format(j+1),fontsize=9)
    plt.show()
```
