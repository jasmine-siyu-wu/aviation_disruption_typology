```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
count = pd.read_csv('final_12.csv').drop('Unnamed: 0',axis=1)
name = ['YYYYMM','DAY','Weekday']
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
for i in huborzone_temp:
    name.append(i+'CX_rate')
    name.append(i+'arr_delay')
    name.append(i+'dep_delay')
    name.append(i+'air_delay')
name.append('labels')
count = count[name]
count
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Generate synthetic 4-dimensional time series data
np.random.seed(42)
time = range(len(count))



# Convert data into a DataFrame
df = count.copy()
df['time'] = time
print(len(df.iloc[0]))

# Initialize and fit the Isolation Forest model
model = IsolationForest(contamination=164/4869, random_state=42)
model.fit(df.iloc[:,3:-2])
print(len(df.iloc[:,3:-2].iloc[0]))

# Predict anomalies
df['anomaly'] = model.predict(df.iloc[:,3:-2])

# -1 for anomaly, 1 for normal
anomalies = df[df['anomaly'] == -1]

# Plot the results for each feature
plt.figure(figsize=(14, 10))

for i, feature in enumerate([' EWRarr_delay']):
    plt.subplot(4, 1, i+1)
    plt.plot(df['time'], df[feature], label=feature)
    plt.scatter(anomalies['time'], anomalies[feature], color='red', label='Anomalies', marker='x')
    plt.title(f'Anomaly Detection in {feature}')
    plt.xlabel('Time')
    plt.ylabel('Value')
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import MinMaxScaler

# Sample Data (randomly generated for demonstration)
#X = np.random.rand(100, 2)

# Initialize the Isolation Forest
iso_forest = IsolationForest(contamination=383/4869, random_state=42)

# Fit the model to the data
iso_forest.fit(df.iloc[:,3:-3])

# Get the raw anomaly score
anomaly_scores = iso_forest.decision_function(df.iloc[:,3:-3])

# Rescale the anomaly scores to be between 0 and 1
scaler = MinMaxScaler()
anomaly_scores_scaled = scaler.fit_transform(anomaly_scores.reshape(-1, 1))

# Output
print("Raw Anomaly Scores:\n", anomaly_scores[:5])  # first 5 scores
print("Scaled Anomaly Scores (0 to 1):\n", anomaly_scores_scaled[:5])  # first 5 scores
```

```python
temp = pd.DataFrame()
temp['labels'] = range(1,13)

temp['descriptions'] = ['West Disturbance','Smooth2','East Super Disruption','ORD Disturbance','NAS Disruption',
                       'Northeast Disruption','DFW Disturbance','NAS Slight Disturbance',
                       'Smooth1','Southeast Disruption','Almost Smooth','East Slight Disturbance']
```

```python
# plt.subplots(figsize=(6.9,4),dpi=300,layout="constrained")
# ax.axvline(y=0.5)
import seaborn as sns

custom_order = [
     'Smooth1','Smooth2', 'Almost Smooth','East Slight Disturbance','NAS Slight Disturbance',
   'DFW Disturbance','ORD Disturbance','West Disturbance',
 'Southeast Disruption','Northeast Disruption', 
    'East Super Disruption',    'NAS Disruption',
      
     
]

# Create a mapping from the sequence
fig, ax = plt.subplots(figsize=(6.9, 4), dpi=300, layout="constrained")

# Add a vertical line (make sure it applies to the x-axis of the boxplot)
ax.axhline(y=np.percentile(df['anomaly_scores_scaled'],95), color='red', alpha = 0.5,linestyle='--', linewidth = 1,label='95 percentile')
ax.axhline(y=np.percentile(df['anomaly_scores_scaled'],99), color='red', alpha = 0.5, linestyle='--',linewidth = 1,label='99 percentile')
order_mapping = {value: index for index, value in enumerate(custom_order)}
data = df.merge(temp,on = 'labels')
data = data.sort_values('descriptions', key=lambda col: col.map(order_mapping))
# Create a boxplot using seaborn
sns.boxplot(x='descriptions', y='anomaly_scores_scaled', data=data)

ax.scatter('ORD Disturbance',0.62,label = 'April 4, 2019',marker = 'x')
ax.scatter('ORD Disturbance',0.07,label = 'May 30, 2018',marker = 'x')
ax.scatter('Southeast Disruption',0.62,label = 'January 2, 2023',marker = 'x')
plt.xticks(rotation=45, fontsize=8)

plt.xlabel('Clusters')
plt.ylabel('Anomaly Score')
plt.legend(fontsize=8)
plt.savefig("../Figures for ATM 25/Figure15.png", dpi=300, bbox_inches='tight');
#plt.yscale('log')
# Show the plot
plt.show()
```
