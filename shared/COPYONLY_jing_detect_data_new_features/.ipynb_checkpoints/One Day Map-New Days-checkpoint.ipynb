```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

## PCA

```python
final = pd.read_csv('newDaysFinalfull.csv').drop('Unnamed: 0',axis=1)
```

```python
airportLocation = pd.read_csv('airportLocation.csv').drop('Unnamed: 0',axis=1)
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
L = [' PHX',' LAX',' SAN',' SFO',' DEN',' FLL',' MIA',' MCO',' TPA',' ATL',
     ' MDW',' ORD',' BWI',' BOS',' DTW',' MSP',' LAS',' EWR',' JFK',' LGA',
     ' CLT',' PHL',' BNA',' AUS',' DFW',' IAH',' SLC',' DCA',' IAD',' SEA',' MCI',' IND',' HNL',' ANC']
```

```python
name = []
for i in huborzone_temp:
    name.append(i+'CX_rate')
    name.append(i+'arr_delay')
    name.append(i+'dep_delay')
    name.append(i+'air_delay')
len(name)
```

```python
hubAirportLocation = airportLocation[airportLocation['LOCID'].isin(L)]
order = list(huborzone_temp)
hubAirportLocation['huborzone'] = pd.Categorical(hubAirportLocation['huborzone'], categories=order, ordered=True)
hubAirportLocation = hubAirportLocation.sort_values(by='huborzone')
```

```python
import matplotlib.cm as cm
x = np.arange(0, 3600, 10) 
y = (x+15*60)/(3600)
cmap = cm.coolwarm
for i in range(len(y)):
    plt.scatter(y[i],np.ones(1),c = cmap(y[i]))
plt.axis('scaled')
plt.ylim(0.99,1.01)
plt.xlim(y.min(),1.25)
plt.yticks([])
plt.xticks([0.25,0.5,0.75,1,1.25],[0,15,30,45,60])
plt.xlabel('minutes');
```

```python
fig, ax = plt.subplots(1, 1,figsize=(2.5,2.5),dpi=300, layout="constrained")


yyyymm = 202408
day = 5


labeledFinal = final[(final['YYYYMM']==yyyymm)&(final['DAY']==day)][name]
cmap = cm.coolwarm
for i in range(len(hubAirportLocation)):
    if hubAirportLocation.iloc[i]['LOCID']==' ORD':
        ax.text(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE']+1, hubAirportLocation.iloc[i]['huborzone'],
        fontsize=4, ha='center', color='black', weight='bold')
    else:
        ax.text(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'], hubAirportLocation.iloc[i]['huborzone'],
        fontsize=4, ha='center', color='black', weight='bold')
    
for i in range(34):
    ax.scatter(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'],
               s = labeledFinal.mean()[i*4]*300,
               c = cmap((labeledFinal.mean()[i*4+1]+15*60)/(3600)))
#ax[0].tick_params(axis='x', labelsize=9)
#ax[0].tick_params(axis='y', labelsize=9)

ax.set_title(yyyymm*100+day,fontsize=10)
ax.set_ylabel('Lat')
ax.set_xlabel('Long')


# for i in range(len(hubAirportLocation)):
#     if hubAirportLocation.iloc[i]['LOCID']==' ORD':
#         ax[1].text(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE']+1, hubAirportLocation.iloc[i]['huborzone'],
#         fontsize=4, ha='center', color='black', weight='bold')
#     else:
#         ax[1].text(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'], hubAirportLocation.iloc[i]['huborzone'],
#         fontsize=4, ha='center', color='black', weight='bold')
    
# for i in range(34):
#     ax[1].scatter(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'],
#                s = labeledFinal.mean()[i*4]*300,
#                c = cmap((labeledFinal.mean()[i*4])*10))
#ax[0].tick_params(axis='x', labelsize=9)
#ax[0].tick_params(axis='y', labelsize=9)

#ax[1].set_title(yyyymm*100+day,fontsize=10)


# Show the plot
plt.show()
```
