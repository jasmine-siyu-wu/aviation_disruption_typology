```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
delay = pd.read_csv('/Users/jing/Downloads/delay.csv',header=None)
delay.columns = ['date','total_delay']
```

```python
delay['year'] = delay.apply(lambda row: datetime.strptime(row.date, '%m/%d/%Y').year,axis=1)
delay['month'] = delay.apply(lambda row: datetime.strptime(row.date, '%m/%d/%Y').month,axis=1)
delay['DAY'] = delay.apply(lambda row: datetime.strptime(row.date, '%m/%d/%Y').day,axis=1)
```

```python
delay['YYYYMM'] = delay['year']*100+delay['month']
```

```python
results = pd.read_csv('final_12.csv').drop('Unnamed: 0',axis=1)
```

```python
results = results.merge(delay,on=['YYYYMM','DAY'])
```

```python
results
```

```python
name = []
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
for i in huborzone_temp:
    name.append(i+'cx')
```

```python
results['cx'] = results[name].sum(axis=1)
```

```python
name = []
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
for i in huborzone_temp:
    results[i+'total_arr_delay'] = results[i+'arr_delay']*results[i+'count']
    name.append(i+'total_arr_delay')
results['total_arr_delay'] = results[name].sum(axis=1)
```

```python
name = []
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
for i in huborzone_temp:
    results[i+'total_dep_delay'] = results[i+'dep_delay']*results[i+'count']
    name.append(i+'total_dep_delay')
results['total_dep_delay'] = results[name].sum(axis=1)
```

```python
name = []
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
for i in huborzone_temp:
    results[i+'total_air_delay'] = results[i+'air_delay']*results[i+'count']
    name.append(i+'total_air_delay')
results['total_air_delay'] = results[name].sum(axis=1)
```

```python
temp = pd.DataFrame()
temp['labels'] = range(12)

temp['descriptions'] = ['West Disturbance','Smooth2','East Super Disruption','ORD Disturbance','NAS Disruption',
                       'Northeast Disruption','DFW Disturbance','NAS Slight Disturbance',
                       'Smooth1','Southeast Disruption','Almost Smooth','East Slight Disturbance']
temp['type'] = [2,1,4,2,4,3,2,1,1,3,1,1]
results = results.merge(temp,on='labels')
```

```python
type_ratio = results.groupby('type').agg({'cx':'sum','total_delay':'sum','date':'count',
                                          'total_arr_delay':'sum','total_dep_delay':'sum',
                                         'total_air_delay':'sum'}).reset_index()
```

```python
type_ratio['cx_ratio'] = type_ratio['cx']/type_ratio['cx'].sum()
type_ratio['total_delay_ratio'] = type_ratio['total_delay']/type_ratio['total_delay'].sum()
type_ratio['date_ratio'] = type_ratio['date']/type_ratio['date'].sum()
type_ratio['arr_delay_ratio'] = type_ratio['total_arr_delay']/type_ratio['total_arr_delay'].sum()
type_ratio['dep_delay_ratio'] = type_ratio['total_dep_delay']/type_ratio['total_dep_delay'].sum()
type_ratio['air_delay_ratio'] = type_ratio['total_air_delay']/type_ratio['total_air_delay'].sum()
```

```python
type_ratio
```

```python
def cdf(a):
    return [a[0],a[0]+a[1],a[0]+a[1]+a[2],1]
def stage(a,c):
    plt.plot([1,1],[0,a[0]],color = c)
    plt.plot([1,2],[a[0],a[0]],color = c)
    plt.plot([2,2],[a[0],a[1]],color = c)
    plt.plot([2,3],[a[1],a[1]],color = c)
    plt.plot([3,3],[a[1],a[2]],color = c)
    plt.plot([3,4],[a[2],a[2]],color = c)
    plt.plot([4,4],[a[2],a[3]],color = c)    
    plt.plot([4,5],[a[3],a[3]],color = c) 
plt.figure(figsize=(20,4))    
stage(cdf(list(type_ratio['cx_ratio'])),'red')    
stage(cdf(list(type_ratio['total_delay_ratio'])),'blue') 
stage(cdf(list(type_ratio['date_ratio'])),'green') 
stage(cdf(list(type_ratio['arr_delay_ratio'])),'orange') 
stage(cdf(list(type_ratio['dep_delay_ratio'])),'pink') 
stage(cdf(list(type_ratio['air_delay_ratio'])),'grey') 
plt.xticks([1.5,2.5,3.5,4.5],['Smooth','Regional Disturbance','Regional Disruption','NAS Disruption'])
```

```python
cluster_ratio = results.groupby('labels').agg({'cx':'sum','total_delay':'sum','date':'count',
                                          'total_arr_delay':'sum','total_dep_delay':'sum',
                                         'total_air_delay':'sum'}).reset_index()
```

```python
cluster_ratio['cx_ratio'] = cluster_ratio['cx']/cluster_ratio['cx'].sum()
cluster_ratio['total_delay_ratio'] = cluster_ratio['total_delay']/cluster_ratio['total_delay'].sum()
cluster_ratio['date_ratio'] = cluster_ratio['date']/cluster_ratio['date'].sum()
cluster_ratio['arr_delay_ratio'] = cluster_ratio['total_arr_delay']/cluster_ratio['total_arr_delay'].sum()
cluster_ratio['dep_delay_ratio'] = cluster_ratio['total_dep_delay']/cluster_ratio['total_dep_delay'].sum()
cluster_ratio['air_delay_ratio'] = cluster_ratio['total_air_delay']/cluster_ratio['total_air_delay'].sum()
```

```python
cluster_ratio['descriptions'] = ['West Disturbance','Smooth2','East Super Disruption','ORD Disturbance','NAS Disruption',
                       'Northeast Disruption','DFW Disturbance','NAS Slight Disturbance',
                       'Smooth1','Southeast Disruption','Almost Smooth','East Slight Disturbance']
custom_order = [
     'Smooth1','Smooth2', 'Almost Smooth','East Slight Disturbance','NAS Slight Disturbance',
   'DFW Disturbance','ORD Disturbance','West Disturbance',
 'Southeast Disruption','Northeast Disruption', 
    'East Super Disruption',    'NAS Disruption',
]
custom_order.reverse()
order_mapping = {value: index for index, value in enumerate(custom_order)}
cluster_ratio = cluster_ratio.sort_values('descriptions', key=lambda col: col.map(order_mapping))
cluster_ratio
```

```python
def cdf(a):
    return [0,a[0],a[0]+a[1],a[0]+a[1]+a[2],a[0]+a[1]+a[2]+a[3],
            a[0]+a[1]+a[2]+a[3]+a[4],a[0]+a[1]+a[2]+a[3]+a[4]+a[5],
            a[0]+a[1]+a[2]+a[3]+a[4]+a[5]+a[6],
            a[0]+a[1]+a[2]+a[3]+a[4]+a[5]+a[6]+a[7],
            a[0]+a[1]+a[2]+a[3]+a[4]+a[5]+a[6]+a[7]+a[8],
            a[0]+a[1]+a[2]+a[3]+a[4]+a[5]+a[6]+a[7]+a[8]+a[9],
            a[0]+a[1]+a[2]+a[3]+a[4]+a[5]+a[6]+a[7]+a[8]+a[9]+a[10],
            a[0]+a[1]+a[2]+a[3]+a[4]+a[5]+a[6]+a[7]+a[8]+a[9]+a[10]+a[11],1]
def stage(a,c,b):
    for i in range(1,13):
        if i == 1:
            plt.plot([i,i],[a[i-1],a[i]],color=c,label=b)
            plt.plot([i,i+1],[a[i],a[i]],color=c)
        else:
            plt.plot([i,i],[a[i-1],a[i]],color=c)
            plt.plot([i,i+1],[a[i],a[i]],color=c)
plt.figure(figsize=(20,4))
stage(cdf(list(cluster_ratio['cx_ratio'])),'red','Cancellations')    
stage(cdf(list(cluster_ratio['total_delay_ratio'])),'blue','OPSNET Delay') 
stage(cdf(list(cluster_ratio['date_ratio'])),'green','# Days') 
stage(cdf(list(cluster_ratio['arr_delay_ratio'])),'orange','Arrival Delay') 
stage(cdf(list(cluster_ratio['dep_delay_ratio'])),'pink','Departure Delay') 
stage(cdf(list(cluster_ratio['air_delay_ratio'])),'grey','Airborne Delay') 
plt.xticks([1.5,2.5,3.5,4.5,5.5,6.5,7.5,8.5,9.5,10.5,11.5,12.5],custom_order,rotation=30,fontsize=12)
plt.yticks(fontsize=12)
plt.ylabel('Cumulative Distribution',fontsize=12)
plt.legend(fontsize=12)
plt.savefig("../Figures for ATM 25/Figure16.png", dpi=300, bbox_inches='tight');
```
