## PCA

```python
final = pd.read_csv('combine.csv').drop('Unnamed: 0',axis=1)
```

```python
airportLocation = pd.read_csv('airportLocation.csv').drop('Unnamed: 0',axis=1)
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
L = [' PHX',' LAX',' SAN',' SFO',' DEN',' FLL',' MIA',' MCO',' TPA',' ATL',
     ' MDW',' ORD',' BWI',' BOS',' DTW',' MSP',' LAS',' EWR',' JFK',' LGA',
     ' CLT',' PHL',' BNA',' AUS',' DFW',' IAH',' SLC',' DCA',' IAD',' SEA',' MCI',' IND',' HNL',' ANC']
```

```python
len(final)
```

```python
name = []
for i in huborzone_temp:
    name.append(i+'CX_rate')
    name.append(i+'arr_delay')
    name.append(i+'dep_delay')
    name.append(i+'air_delay')
len(name)
```

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Standardize the data
scaler = StandardScaler()
x_scaled = scaler.fit_transform(final[name])

# Apply PCA
pca = PCA(n_components=30, random_state=42)
principal_components = pca.fit_transform(x_scaled)

# Create DataFrame for the principal components
pca_df = pd.DataFrame(data=principal_components)


# Print the explained variance
explained_variance = pca.explained_variance_ratio_
explained_variance_sum = sum(explained_variance)
print(explained_variance_sum,f'Explained Variance: {explained_variance}')

loadings = pca.components_

# Create a DataFrame for better readability
loadings_df = pd.DataFrame(loadings, )
```

```python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data
np.random.seed(42)
X = np.array(pca_df.iloc[:,:24])
#X = np.random.rand(100, 2)

# Perform KMeans clustering
kmeans = KMeans(n_clusters=12, random_state=42)
kmeans.fit(X)

# Get labels and cluster centers
labels = kmeans.labels_
centers = kmeans.cluster_centers_
```

```python
from sklearn.metrics import pairwise_distances_argmin_min
closest, _ = pairwise_distances_argmin_min(centers, pca_df.iloc[:,:24])
```

```python
import numpy as np
from sklearn.cluster import KMeans



# Apply K-Means clustering
n_clusters = 12

labels = kmeans.fit_predict(X)
centroids = kmeans.cluster_centers_

# Compute concentration scores (mean distance to centroid)
concentration_scores = []
for k in range(n_clusters):
    cluster_points = X[labels == k]
    centroid = centroids[k]
    avg_distance = np.mean(np.linalg.norm(cluster_points - centroid, axis=1))
    concentration_scores.append(avg_distance)

# Print concentration scores for each cluster
for i, score in enumerate(concentration_scores):
    print(f"Cluster {i}: Concentration Score = {score:.4f}")
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize

# Example time series cluster labels
clusters = np.array(final['labels'])  # Replace with your data

# Number of clusters
n_clusters = 12

# Initialize the transition matrix
transition_matrix = np.zeros((n_clusters, n_clusters))

# Calculate the transition counts
for (i, j) in zip(clusters[:-1], clusters[1:]):
    transition_matrix[i, j] += 1

# Normalize to get probabilities
transition_matrix = normalize(transition_matrix, axis=1, norm='l1')

# Plotting the heatmap
plt.figure(figsize=(20, 16))
sns.heatmap(transition_matrix, annot=True, fmt=".2%",cmap='PiYG', center=0.1,cbar=True,
            xticklabels=[f'Cluster {i+1}' for i in range(n_clusters)],
            yticklabels=[f'Cluster {i+1}' for i in range(n_clusters)])
plt.title("Transition Matrix Heatmap")
plt.xlabel("To Cluster")
plt.ylabel("From Cluster")
plt.show()
```
