```python
temp = []
days = 0
temp1 = 0
worst_day = 0
worst_day_score = 0
i = 0
while i < 4869:
    if full.iloc[i]['score']==1:
        if days == 0:
            temp1=full.iloc[i]['YYYYMM']*100+full.iloc[i]['DAY']
        days = days+1
        if worst_day_score < full.iloc[i]['anomaly_scores_scaled']:
            worst_day_score = full.iloc[i]['anomaly_scores_scaled']
            worst_day = full.iloc[i]['YYYYMM']*100+full.iloc[i]['DAY']
    else:
        if temp1 != 0:
            temp.append([temp1,days,worst_day,worst_day_score])
        temp1 = 0
        worst_day = 0
        days = 0
        worst_day_score = 0
    i = i+1
        
```

```python
temp = pd.DataFrame(temp)
```

```python
temp.columns = ['StartDay','Duration','WorstDay','WorstDayScore']
```

```python
temp['StartYYYYMM'] = temp['StartDay']//100
temp['StartDay'] = temp['StartDay']%100

temp['WorstYYYYMM'] = temp['WorstDay']//100
temp['WorstDay'] = temp['WorstDay']%100
```

```python
temp['StartYYYYMM'] = temp.apply(lambda row: int(row.StartYYYYMM),axis=1)
temp['StartDay'] = temp.apply(lambda row: int(row.StartDay),axis=1)
temp['WorstYYYYMM'] = temp.apply(lambda row: int(row.WorstYYYYMM),axis=1)
temp['WorstDay'] = temp.apply(lambda row: int(row.WorstDay),axis=1)
```
