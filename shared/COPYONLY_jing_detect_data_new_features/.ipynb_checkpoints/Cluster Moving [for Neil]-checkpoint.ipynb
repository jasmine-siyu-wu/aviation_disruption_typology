```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from matplotlib.colors import Normalize
from matplotlib import cm
from matplotlib.colorbar import ColorbarBase
```

```python
results = pd.read_csv('final_12.csv').drop('Unnamed: 0',axis=1)
```

```python
labels = pd.DataFrame()
labels['labels'] = range(12)
labels['description'] = ['West Disturbance','Smooth2','East Super Disruption','ORD','NAS Disruption',
                         'Northeast','DFW','NAS Slightly','Smooth1','Southeast','Almost Smooth','East Slightly']
labels['4 groups'] = [2,1,4,2,4,3,2,1,1,3,1,1]
```

```python
results = results.merge(labels,on='labels')
```

```python
results = results[['YYYYMM','DAY','Weekday','labels','description','4 groups']]
```

```python
final = pd.read_csv('combine.csv').drop('Unnamed: 0',axis=1)
airportLocation = pd.read_csv('airportLocation.csv').drop('Unnamed: 0',axis=1)
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
L = [' PHX',' LAX',' SAN',' SFO',' DEN',' FLL',' MIA',' MCO',' TPA',' ATL',
     ' MDW',' ORD',' BWI',' BOS',' DTW',' MSP',' LAS',' EWR',' JFK',' LGA',
     ' CLT',' PHL',' BNA',' AUS',' DFW',' IAH',' SLC',' DCA',' IAD',' SEA',' MCI',' IND',' HNL',' ANC']
name = []
for i in huborzone_temp:
    name.append(i+'CX_rate')
    name.append(i+'arr_delay')
    name.append(i+'dep_delay')
    name.append(i+'air_delay')
hubAirportLocation = airportLocation[airportLocation['LOCID'].isin(L)]
order = list(huborzone_temp)
hubAirportLocation['huborzone'] = pd.Categorical(hubAirportLocation['huborzone'], categories=order, ordered=True)
hubAirportLocation = hubAirportLocation.sort_values(by='huborzone')
```

```python
for i in range(len(results)):
    if results.iloc[i]['labels']==4:
        j=1
        while results.iloc[i-j]['4 groups']!=1:
            j=j+1
        k=1
        while results.iloc[i+k]['4 groups']!=1:
            k=k+1
        print(results.iloc[i-j]['YYYYMM'],results.iloc[i-j]['DAY'])
        print(list(results.iloc[i-j:i+k+1]['description']))
        print(list(results.iloc[i-j:i+k+1]['4 groups']))
        print('\t')

     
```

```python
yyyymm = 202403
day = 9


labeledFinal = final[(final['YYYYMM']==yyyymm)&(final['DAY']==day)][name]
cmap = cm.coolwarm
fig, ax = plt.subplots(figsize=(4, 4),dpi=100)
for i in range(len(hubAirportLocation)):
    ax.text(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'], hubAirportLocation.iloc[i]['LOCID'],
        fontsize=6, ha='center', color='black', weight='bold')
for i in range(34):
    plt.scatter(hubAirportLocation.iloc[i]['LONGITUDE'],hubAirportLocation.iloc[i]['LATITUDE'],
               s = labeledFinal.mean()[i*4]*300,
               c = cmap((labeledFinal.mean()[i*4+1]+15*60)/(3600)))
ax.tick_params(axis='x', labelsize=9)
ax.tick_params(axis='y', labelsize=9)
plt.xlabel('Lon',fontsize = 9)
plt.ylabel('Lat',fontsize = 9)
plt.title(yyyymm*100+day)
plt.show()
```
