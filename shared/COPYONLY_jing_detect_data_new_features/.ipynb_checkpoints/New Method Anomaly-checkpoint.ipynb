```python
count = pd.read_csv('final_12.csv').drop('Unnamed: 0',axis=1)
name = ['YYYYMM','DAY','Weekday']
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
for i in huborzone_temp:
    name.append(i+'CX_rate')
    name.append(i+'arr_delay')
    name.append(i+'dep_delay')
    name.append(i+'air_delay')
name.append('labels')
count = count[name]
count
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Generate synthetic 4-dimensional time series data
np.random.seed(42)
time = range(len(count))



# Convert data into a DataFrame
df = count.copy()
df['time'] = time
print(len(df.iloc[0]))

# Initialize and fit the Isolation Forest model
model = IsolationForest(contamination=487/4869, random_state=42)
model.fit(df.iloc[:,3:-2])
print(len(df.iloc[:,3:-2].iloc[0]))

# Predict anomalies
df['anomaly'] = model.predict(df.iloc[:,3:-2])

# -1 for anomaly, 1 for normal
anomalies = df[df['anomaly'] == -1]

# Plot the results for each feature
plt.figure(figsize=(14, 10))

for i, feature in enumerate([' EWRarr_delay']):
    plt.subplot(4, 1, i+1)
    plt.plot(df['time'], df[feature], label=feature)
    plt.scatter(anomalies['time'], anomalies[feature], color='red', label='Anomalies', marker='x')
    plt.title(f'Anomaly Detection in {feature}')
    plt.xlabel('Time')
    plt.ylabel('Value')
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import MinMaxScaler

# Sample Data (randomly generated for demonstration)
#X = np.random.rand(100, 2)

# Initialize the Isolation Forest
iso_forest = IsolationForest(contamination=383/4869, random_state=42)

# Fit the model to the data
iso_forest.fit(df.iloc[:,3:-3])

# Get the raw anomaly score
anomaly_scores = iso_forest.decision_function(df.iloc[:,3:-3])

# Rescale the anomaly scores to be between 0 and 1
scaler = MinMaxScaler()
anomaly_scores_scaled = scaler.fit_transform(anomaly_scores.reshape(-1, 1))

# Output
print("Raw Anomaly Scores:\n", anomaly_scores[:5])  # first 5 scores
print("Scaled Anomaly Scores (0 to 1):\n", anomaly_scores_scaled[:5])  # first 5 scores
```

```python
import numpy as np

# t = estimator.tree_
# left, right, feat, thr = t.children_left, t.children_right, t.feature, t.threshold
# N = t.node_count

# # Build parent[] and depth[] once
# parent = np.full(N, -1, dtype=int)   # root has parent -1
# depth  = np.zeros(N, dtype=int)
# stack = [(0, 0)]
# while stack:
#     node, d = stack.pop()
#     depth[node] = d
#     for child in (left[node], right[node]):
#         if child != -1:
#             parent[child] = node
#             stack.append((child, d + 1))

def lca(a, b):
    """Lowest common ancestor of nodes a and b."""
    # Raise deeper node up to the same depth
    while depth[a] > depth[b]:
        a = parent[a]
    while depth[b] > depth[a]:
        b = parent[b]
    # Climb together until they meet
    while a != b:
        a = parent[a]
        b = parent[b]
    return a  # index of the LCA node

# Optional helpers:
def split_rule(node_idx):
    """Readable split rule for a node (the ancestor)."""
    f = feat[node_idx]
    th = thr[node_idx]
    name = getattr(estimator, "feature_names_in_", None)
    fname = (name[f] if name is not None and f >= 0 else f"f{f}")
    return f"{fname} <= {th:.6f}"

def path_to_root(node_idx):
    """List of (ancestor, is_left_branch) from root->node."""
    path = []
    while parent[node_idx] != -1:
        p = parent[node_idx]
        is_left = (left[p] == node_idx)
        path.append((p, is_left))
        node_idx = p
    return list(reversed(path))
```

```python
scores1 = []
temp = df[df['anomaly'] == True]
for k in range(100):
    print(k)
    estimator = model.estimators_[k]
    t = estimator.tree_
    left, right, feat, thr = t.children_left, t.children_right, t.feature, t.threshold
    N = t.node_count

    parent = np.full(N, -1, dtype=int)   # root has parent -1
    depth  = np.zeros(N, dtype=int)
    stack = [(0, 0)]
    while stack:
        node, d = stack.pop()
        depth[node] = d
        for child in (left[node], right[node]):
            if child != -1:
                parent[child] = node
                stack.append((child, d + 1))
    
    for i in range(len(temp)-1):
        for j in range(i+1,len(temp)):
            estimator = model.estimators_[k]
            score = lca(estimator.apply(np.array(temp.iloc[i, 3:-5]).reshape(1, -1))[0], estimator.apply(np.array(temp.iloc[j, 3:-5]).reshape(1, -1))[0])
            scores1.append([i,j,k,depth[score]])
```

```python
scores_table = pd.DataFrame(scores1,columns = ['nodei','nodej','treek','similar'])
```

```python
scores_table['similar'].value_counts()
```

```python
scores_table['distance'] = 8-scores_table['similar']
```

```python
scores_table = scores_table.groupby(['nodei','nodej']).mean()[['distance']].reset_index()
```

```python
scores = np.array(scores_table)
n = int(scores[:, :2].max() + 1)
D = np.zeros((n, n))

for i, j, d in scores:
    D[int(i), int(j)] = d   # ✅ square brackets
    D[int(j), int(i)] = d

np.fill_diagonal(D, 0)
```

```python
D
```

```python
from sklearn.cluster import SpectralClustering
import numpy as np

# Convert distance to similarity (kernel)
sigma = np.median(D)
S = np.exp(-D**2 / (2 * sigma**2))

clustering = SpectralClustering(n_clusters=5, affinity='precomputed')
labels1 = clustering.fit_predict(S)
```
