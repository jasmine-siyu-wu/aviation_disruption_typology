```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

```python
dep_delay = pd.read_csv('newDaysFinalDEP.csv').drop('Unnamed: 0',axis=1)
```

```python
arr_delay = pd.read_csv('newDaysFinalARR.csv').drop('Unnamed: 0',axis=1)
```

```python
arr_cx = pd.read_csv('Cancelled_34_24_25.csv').drop('Unnamed: 0',axis=1)
```

```python
dep_cx = pd.read_csv('Cancelled_34_DEP_24_25.csv').drop('Unnamed: 0',axis=1)
```

```python
huborzone_temp = np.array([' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA'])
```

```python
names = ['YYYYMM','DAY']
for i in huborzone_temp:
    names.append(i+'count')
    names.append(i+'arr_delay')
    names.append(i+'dep_delay')
    names.append(i+'DELAY_AIR')
dep_delay.columns = names
arr_delay.columns = names
```

```python
names = ['YYYYMM','DAY']
for i in huborzone_temp:
    names.append(i+'cx_count')
arr_cx.columns = names
dep_cx.columns = names
```

```python
#dep_delay = dep_delay[dep_delay['YYYYMM']<=202409]
#arr_delay = arr_delay[arr_delay['YYYYMM']<=202409]
```

```python
full = pd.DataFrame()
full['YYYYMM'] = arr_delay['YYYYMM']
full['DAY'] = arr_delay['DAY']
for i in huborzone_temp:
    full[i+'arr_delay'] = arr_delay[i+'arr_delay']
    full[i+'air_delay'] = arr_delay[i+'DELAY_AIR']
    full[i+'dep_delay'] = dep_delay[i+'dep_delay']
    full[i+'count'] = arr_delay[i+'count']+dep_delay[i+'count']
    full[i+'cx'] = arr_cx[i+'cx_count']+dep_cx[i+'cx_count']
    full[i+'CX_rate'] = full[i+'cx']/(full[i+'cx']+full[i+'count'])
```

```python
full.to_csv('newDaysFinalfull_24_25.csv')
```
