```python
flight = pd.read_csv('/Users/jing/Documents/Flight Data/DATA/ASPM2024-2025.8.CSV.gz')
```

```python
a = ['PABE',' GPI','PAGA',' DGG',' SGJ',' HEF',' PFN']
b = [' BET',' FCA',' GAL',' RDB',' UST',' MNZ',' ECP']
flight = flight.replace(a,b)
```

```python
flight = flight[flight['OAG']=='Y']
```

```python
flight = flight[flight['USER_CLASS']=='C']
```

```python
flight = flight[flight['FLT_TYPE']=='N']
```

```python
flight = flight[(flight['DEP_LOCID'].isin(Airports_temp))&(flight['ARR_LOCID'].isin(Airports_temp))]
```

```python
airport = pd.read_csv('airportHuborZone.csv').drop('Unnamed: 0',axis=1)
```

```python
#DEP side
```

```python
flight = flight.merge(airport,on = 'DEP_LOCID')
```

```python
flight['dep_delay'] = flight.apply(lambda row: max(0, row.OOOI_DEP-row.OAG_S_DEP),axis=1)
flight['arr_delay'] = flight.apply(lambda row: max(0, row.OOOI_ARR-row.OAG_S_ARR),axis=1)
```

```python
flight
```

```python
count = flight.groupby(['DEP_YYYYMM','DEP_DAY','huborzone']).agg({'DEP_QTR':'count','arr_delay':'mean','dep_delay':'mean','DELAY_AIR':'mean'}).reset_index()
```

```python
count
```

```python
count1 = flight.groupby(['DEP_YYYYMM','DEP_DAY']).mean().reset_index()
count1 = count1[['DEP_YYYYMM','DEP_DAY']]
count1
```

```python
huborzone_temp = [' EWR', 'ZFW', 'ZAB', 'ZMP', ' TPA', 'ZBW', 'ZSE', ' PHL', 'ZAN',
       ' DCA', ' IAH', ' CLT', ' ATL', 'ZDV', ' ORD', 'ZOB', ' MIA',
       ' AUS', ' MDW', ' LAX', ' LAS', 'ZOA', 'ZLC', 'ZME', 'ZID', 'ZKC',
       ' MCO', ' BWI', ' IAD', ' SAN', ' FLL', 'ZHN', ' JFK', ' LGA']
```

```python
empty = []
for i in np.unique(count[['DEP_YYYYMM','DEP_DAY']], axis=0):
    for j in range(len(huborzone_temp)):
        empty.append([i[0],i[1],huborzone_temp[j]])
empty = pd.DataFrame(empty,columns = ['DEP_YYYYMM','DEP_DAY','huborzone'])
len(empty)#34*4869
```

```python
full = empty.merge(count,on = ['DEP_YYYYMM','DEP_DAY','huborzone'],how = 'left')
full = full.fillna(0)
full
```

```python
final = []
for i in range(len(count1)):
    temp = []
    temp.append(count1.iloc[i]['DEP_YYYYMM'])
    temp.append(count1.iloc[i]['DEP_DAY'])
    for j in range(34):
        temp.append(full.iloc[i*34+j]['DEP_QTR'])
        temp.append(full.iloc[i*34+j]['arr_delay'])
        temp.append(full.iloc[i*34+j]['dep_delay'])
        temp.append(full.iloc[i*34+j]['DELAY_AIR'])
    final.append(temp)
```

```python
final = pd.DataFrame(final)
final
```

```python
final.to_csv('newDaysFinalDEP.csv')
```
